## SVM to predict customer will buy SUV for not

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Social_Network_Ads.csv')
data.head()
```

```python
# remove userID column
data = data.drop('User ID', axis= 1)
data.head()
```

```python
X = data.iloc[:, :-1].values
Y = data.iloc[:, -1].values

print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=0)

print(X_train.shape)
print(X_test.shape)
```

```python
print(X_train)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

CT = ColumnTransformer(
    transformers = [
        ('encoder', OneHotEncoder(drop='first', sparse=False), [0]),
        ('scaler', StandardScaler(), [1,2])
    ]
)
```

```python
# Fit it in X_train and transform both X_train and X_test
X_train = CT.fit_transform(X_train)
X_test = CT.transform(X_test)
```

```python
print(X_train)
```

### Train SVM Model

```python
from sklearn.svm import SVC
classifier = SVC(kernel='linear', random_state=0)
classifier.fit(X_train, Y_train)
```

```python
y_pred = classifier.predict(X_test)
print(y_pred)
```

```python
result = np.concatenate(
    (y_pred.reshape(len(y_pred), 1), Y_test.reshape(len(Y_test), 1)),
    axis = 1
)
print(result)
```

### Model evaluation

```python
from sklearn.metrics import confusion_matrix, accuracy_score
print('Confusion matrix\n',confusion_matrix(Y_test, y_pred))
print('\nAccuracy Score\n',accuracy_score(Y_test, y_pred))
```

### Predict new data

```python
new_data = np.array([['Female', 27, 97000]], dtype=object)
print(new_data)
```

```python
transformed_data = CT.transform(new_data)
print(transformed_data)
```

```python
result = classifier.predict(transformed_data)
print(f'Model predicted {result}.')
```

## Kernel SVM for non-linear data

```python
from sklearn.svm import SVC
classifier2 = SVC(kernel='rbf', random_state=0)
classifier2.fit(X_train, Y_train)
```

```python
y_pred2 = classifier2.predict(X_test)
print(y_pred2)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
print('Confusion matrix\n',confusion_matrix(Y_test, y_pred2))
print('\nAccuracy Score\n',accuracy_score(Y_test, y_pred2))
```
