## Logistic regression
- Linear classification model<br><br>
Problem: Company XYZ has launched a new SUV and based on prev data, manager wants to predict if a customer will buy this new SUV or not.<br>
Here we have age and estimated salary of customer as independent variables. Purchase is target/dependent varaible with binary value 0 and 1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Social_Network_Ads.csv')
data.head()
```

```python
# drop unnecessary column - giving no info
data = data.drop('User ID', axis=1)
data.head()
```

```python
X = data.iloc[:,:-1].values
Y = data.iloc[:,-1].values

print(X.shape)
print(Y.shape)
```

```python
print(X)
```

### check missing values

```python
print(data.isnull().sum())
```

### Splitting Dataset

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=0)
```

```python
print(X_train)
```

```python
print(X_train.shape)
print(X_test.shape)
```

### Categorical column encoding

```python
data.info()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

CT = ColumnTransformer(
    transformers = [
        ('encoder', OneHotEncoder(drop='first', sparse=False), [0])
    ],
    remainder = 'passthrough'
)
```

fit_transform done on training data, while transform done on test dataset

```python
X_train = CT.fit_transform(X_train)
X_test = CT.transform(X_test)
```

```python
print(X_train)
```

### Feature scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

Scale all column except first, as it is category column and scaling it will loose it's meaning.<br>
Also apply fit_transform on X_train and only on X_test to avoid data leakage.

```python
X_train[:, 1:] = scaler.fit_transform(X_train[:, 1:])
X_test[:, 1:] = scaler.transform(X_test[:, 1:])
```

```python
print(X_train)
```

### Training Logistic Regression model

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression()
classifier.fit(X_train, Y_train)
```

```python
# predicting test set results
y_pred = classifier.predict(X_test)
print(y_pred)
```

### Predict new data
- predict if a female customer of age 27 with salary 97000 will buy the new SUV car or not

- new data must be passed in same way as it is trained

```python
# new_data = pd.DataFrame(
#     [['Female', 27, 97000]],
#     columns=['Gender', 'Age', 'EstimatedSalary']  # must match your training columns
# )
# print(new_data)

new_data = np.array([['Female', 27, 97000]], dtype=object)
print(new_data)
```

```python
new_data_encoded = CT.transform(new_data)
print(new_data_encoded)
```

```python
# scale it
scaled_data = scaler.transform(new_data_encoded[:,1:])
print(scaled_data)
```

```python
final_data = np.concatenate(
    [new_data_encoded[:,:1], scaled_data], axis=1
)
print(final_data)
```

```python
result = classifier.predict(final_data)
print(f'Model predicted {result}. Customer will not buy SUV')
```
