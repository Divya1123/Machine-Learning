## Simple Linear Regression

In simple linear regression data we have 1 feature and 1 target. Here we have data having year of experience as feature and we are predicting their salary.

### Import Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Salary_Data.csv')
data.head()
```

```python
X = data.iloc[:, :-1].values
Y = data.iloc[:, -1].values

print(X)
print('\n', Y)
```

### Checking missing data

```python
print(data.isnull().sum())
```

## Splitting Data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 0)
```

```python
print(len(X_train))
print(len(X_test))
```

### Training Simple Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor = LinearRegression()          # created instance of LinearRegression Model
regressor.fit(X_train, Y_train)         # trained the model with training data
```

```python
# Coefficient (slope/s)
print("Coefficients:", regressor.coef_)

# Intercept
print("Intercept:", regressor.intercept_)

# regression equation
print(f"Equation: y = {regressor.intercept_} + {regressor.coef_[0]} * x")
```

### Predicting test set results

```python
y_pred = regressor.predict(X_test)
```

### Visualize training set results<br>
Red represent actual dataset (X_train, Y_train) and blue represent predicted training values<br>
scatter for scatter plot and plot for regression line.<br>

```python
plt.scatter(X_train, Y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Salary vs Experience (Training Set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

### Visualize Test Set Results<br>
Red represent actual dataset (X_test, Y_test) and blue represent predicted training values(y_test, y_pred)<br>
Scatter plot visualize actual test data point (X_test, Y_test).
Plot used to plot regression line learned on training data.<br>
Here we will not change the plot regression line, bec regression line calculated on training set is used to predict test values. So regression line plot will be on training data.<br><br>

plot blue line is made up of predicted values.

```python
plt.scatter(X_test, Y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Salary vs Experience (Training Set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```
