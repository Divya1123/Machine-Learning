### Project 
Here we have dataset of retail company which consist of customer country, age, salary and if they purchased their product or not.

### Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Importing the Dataset

- iloc for slicing
- using .values we are returning numpy array instead of dataframe by removing col name and index.
<br> Some ML libraries uses numpy instead of df

```python
# creating dataframe named dataset
dataset = pd.read_csv('Data.csv')

X = dataset.iloc[:, :-1].values
Y = dataset.iloc[:, -1].values
```

```python
print(X)
print('\n',Y)
```

### Handling Missing data
- Missing data can cause error while training the ML models
- we can use scikit-learn
- If we are using mean or median as strategy, then imputer is applied to column with numeric values. While if we are using most_frequent or constant, we can apply to both numeric and categorical.

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

# fit imputer to calculate mean of sec and third column
imputer.fit(X[ :, 1:3])

# apply and replace the calculated value in data             
X[ :, 1:3] = imputer.transform(X[ :, 1:3])

print(X)
```

### Encoding categorical Data

1- Encoding Independent variable


```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(
    transformers = [ 
        ('encoder', OneHotEncoder(drop='first'), [0])
        ],
    remainder = 'passthrough'
)
```

```python
X = ct.fit_transform(X)
print(type(X))
print('\n',X)
```

2- Encoding Target/Dependent Variable

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
Y = encoder.fit_transform(Y)
```

```python
print(Y)
```

### Splitting data into test train dataset

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=1)
```

```python
print(X_train)
print('\n',X_test)
print('\n',Y_train)
print('\n',Y_test)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train[:, 2:] = scaler.fit_transform(X_train[:, 2:])
X_test[:, 2:] = scaler.transform(X_test[:, 2:])
```

```python
print(X_train)
print('\n', X_test)
```
