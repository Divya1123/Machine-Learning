## Normalization

Way 1 - Using Pure Python

```python
# sample column
data = [10,20,30,40]

# find min and max value
min_val = min(data)
max_val = max(data)

# Normalize each value
normalized = [(x-min_val)/(max_val-min_val) for x in data]

print(normalized)
```

Way 2 - Using Pandas

```python
import pandas as pd
```

```python
# Sample dataframe
df = pd.DataFrame({'values': [10,20,30,40]})

# Min-Max Normalization on single column
df['normalized'] = (df['values']-df['values'].min())/(df['values'].max() - df['values'].min())

print(df)
```

```python

# Example DataFrame
df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [100, 200, 300, 400, 500]
})

# Apply min-max normalization to all columns
df_normalized = (df - df.min()) / (df.max() - df.min())

print("Original:\n", df)
print("\nNormalized:\n", df_normalized)
```

Way 3 - Using Scikit-Learn

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
# Sample Dataframe
df = pd.DataFrame({
    'A': [10,20,30,40],
    'B': [5,10,15,25]
})

# create scaler
scaler = MinMaxScaler()

# fit the scaler to the data (learn min and max of each column)
scaler.fit(df)
print('min values of each column:', scaler.data_min_)
print('max values of each column:', scaler.data_max_)

# trasform the data (apply normalization)
scaled_array = scaler.transform(df)
print('\nScaled array:', scaled_array)

# convert array to df
df_scaled = pd.DataFrame(scaled_array, columns = df.columns)
print('\n', df_scaled)
```

## Standardization

Way 1 - Using Pure Python

```python
# Sample dat
data = [10,20,30,40]

# compute mean
mean = sum(data)/len(data)

# compute standard deviation
std = ( sum((x-mean)**2 for x in data)/ len(data) ) ** 0.5

# standardize data
standardized_data = [(x-mean)/std for x in data]

print("Original data:", data)
print("Standardized data:", standardized_data)
```

Way 2 - using pandas

[we can also do this on numpy as well]

```python
# sample dataframe
df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [100, 200, 300, 400, 500]
})

# standardization
df_standardized = (df-df.mean()) / df.std()

print("Original DataFrame:\n", df)
print("\nStandardized DataFrame:\n", df_standardized)
```

Way 3 - Using scikit-learn

```python
from sklearn.preprocessing import StandardScaler

df = df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [100, 200, 300, 400, 500]
})

scaler = StandardScaler()

scaled_array = scaler.fit_transform(df)
print(scaled_array)

df_standardized = pd.DataFrame(scaled_array, columns = df.columns)
print(df_standardized)
```
